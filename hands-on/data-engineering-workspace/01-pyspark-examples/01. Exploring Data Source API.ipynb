# Dealing with Bad Records

1. PERMISSIVE (Default) : The bad record will be inserted into _corrupt_record column
2. DROPMALFORMED : Drop the bad record
3. FAILFAST : Fails to create DF

```python
spark.read.json(
    path="/Volumes/walmart16112026/quickstart_schema/sandbox/dataset/access_logs.json",
    mode="FAILFAST"
).display()
```
